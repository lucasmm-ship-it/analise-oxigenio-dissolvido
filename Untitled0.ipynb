```python

import pandas as pd
import requests

url = "https://www.snirh.gov.br/arcgis/rest/services/INDE/Camadas/MapServer/137/query"

todos_os_dados = []
offset = 0

while True:
    parametros = {
        "where": "1=1",
        "outFields": "*",
        "returnGeometry": "false",
        "f": "json",
        "resultOffset": offset,
        "resultRecordCount": 2000
    }

    resposta = requests.get(url, params=parametros)
    resposta.raise_for_status()
    resultado = resposta.json()

    registros = resultado.get("features", [])

    if not registros:
        break

    todos_os_dados.extend(
        [registro["attributes"] for registro in registros]
    )

    offset += len(registros)

dados = pd.DataFrame(todos_os_dados)

print("Total de registros:", len(dados))
display(dados.head())

dados.to_csv("dados_oxigenio_dissolvido.csv", index=False)
```

```python

# 1. Dimensões da base
print("Quantidade de estações e colunas:")
print(dados.shape)

# 2. Informações sobre os dados
print("\nInformações da base:")
dados.info()

# 3. Quantidade de estações por estado
print("\nEstações por estado:")
print(dados["sguf"].value_counts())

# 4. Valores ausentes
print("\nValores ausentes:")
print(dados.isnull().sum().sort_values(ascending=False).head(15))
```

```python

# Verificar os dados de oxigênio dissolvido entre 2001 e 2020

anos = [f"med_{ano}" for ano in range(2001, 2021)]

# Verificar quais colunas existem
colunas_disponiveis = [col for col in anos if col in dados.columns]

print("Anos disponíveis:", colunas_disponiveis)

# Quantidade de valores disponíveis por ano
print("\nQuantidade de registros por ano:")
print(dados[colunas_disponiveis].count())

# Média geral de OD por ano
print("\nMédia de oxigênio dissolvido por ano:")
print(dados[colunas_disponiveis].mean().round(2))
```

```python

# Separar as estações do Maranhão
dados_ma = dados[dados["sguf"] == "MA"].copy()

print("Estações do Maranhão:", len(dados_ma))

# Quantidade de medições disponíveis por ano no Maranhão
print("\nMedições disponíveis no Maranhão:")
print(dados_ma[colunas_disponiveis].count())

# Médias anuais de oxigênio dissolvido no Maranhão
print("\nMédias anuais de OD no Maranhão:")
print(dados_ma[colunas_disponiveis].mean().round(2))
```

```python

# Separar os dados de 2020 em dois grupos

ma_2020 = dados.loc[
    dados["sguf"] == "MA", "med_2020"
].dropna()

outros_2020 = dados.loc[
    dados["sguf"] != "MA", "med_2020"
].dropna()

# Estatísticas descritivas
print("MARANHÃO")
print("Número de observações:", len(ma_2020))
print("Média:", round(ma_2020.mean(), 2))
print("Mediana:", round(ma_2020.median(), 2))
print("Desvio padrão:", round(ma_2020.std(), 2))

print("\nDEMAIS ESTADOS")
print("Número de observações:", len(outros_2020))
print("Média:", round(outros_2020.mean(), 2))
print("Mediana:", round(outros_2020.median(), 2))
print("Desvio padrão:", round(outros_2020.std(), 2))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.boxplot(
    [ma_2020, outros_2020],
    labels=["Maranhão", "Demais UFs"]
)

plt.title("Oxigênio dissolvido por grupo em 2020")
plt.ylabel("Média de oxigênio dissolvido")
plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
print("MARANHÃO")
print(ma_2020.describe())

print("\nDEMAIS UFs")
print(outros_2020.describe())

print("\nVALORES EXTREMOS")
print("Maranhão:", ma_2020.min(), "até", ma_2020.max())
print("Demais UFs:", outros_2020.min(), "até", outros_2020.max())
```

```python
from scipy.stats import ttest_ind

resultado = ttest_ind(
    ma_2020,
    outros_2020,
    equal_var=False,
    alternative="two-sided"
)

print("Estatística t:", resultado.statistic)
print("Valor-p:", resultado.pvalue)

if resultado.pvalue < 0.05:
    print("Resultado: rejeitamos H0.")
    print("Existe diferença estatisticamente significativa entre as médias.")
else:
    print("Resultado: não rejeitamos H0.")
    print("Não há evidências suficientes de diferença entre as médias.")
```
